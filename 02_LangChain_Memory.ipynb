```python
## LangChain会话记忆管理
import os
# 获取api_key
api_key = os.environ.get("DEEPSEEK_API_KEY")
```

```python
## 导包,此处使用的是新版本的LangChain包
from langchain_openai.chat_models import ChatOpenAI
from langchain_classic.chains import ConversationChain
from langchain_classic.memory import ConversationBufferMemory
```

```python
llm = ChatOpenAI(
    temperature=0,
    base_url="https://api.deepseek.com",
    model="deepseek-v4-flash",
    api_key=api_key
)
memory = ConversationBufferMemory()
conversations = ConversationChain( # 创建一个对话链
    llm=llm,
    memory=memory,
    verbose=True # 将verbose设置为true可以了解更多的细节
)
```

```python
# 输入函数
conversations.predict(input="Hi, my name is jasmine")
```

```python
conversations.predict(input="What is 1+1?")
```

```python
conversations.predict(input="What is my name?")
```

```python
# 使用变量memory保存会话记忆
print(memory.buffer)
```

```python
memory.load_memory_variables({}) # 目前传入一个空字典
```

```python
# 可以向ConversationBufferMemory当中加入输入和输出
memory = ConversationBufferMemory()
memory.save_context({"input":"Hi"},{"output":"What's up"}) # 加入输入、输出
```

```python
print(memory.buffer)
```

```python
memory.load_memory_variables({})
```

```python
memory.save_context({"input":"not much, just hanging"},
                    {"output": "Cool"})

# 大模型本身是无状态的，每个请求交互调用的api都是独立的
```

```python
conversations.predict(input="what's my name")
```

```python
# 另一种会话记忆方式: 保留窗口记忆，保留最后若干轮对话信息
# 可以防止记忆存储量随对话的进行而无限增长
from langchain_classic.memory import ConversationBufferWindowMemory
```

```python
memory = ConversationBufferWindowMemory(k=1) # 只希望记住发出的最后一条对话
```

```python
memory.save_context({"input":"Hi"},{"output":"What's up"}) # 加入输入、输出
memory.save_context({"input":"not much, just hanging"},
                    {"output": "Cool"})
```

```python
memory.load_memory_variables({})
```

```python
llm = ChatOpenAI(
    temperature=0,
    base_url="https://api.deepseek.com",
    model="deepseek-v4-flash",
    api_key=api_key
)
memory = ConversationBufferWindowMemory(k=1)
conversations = ConversationChain(
    llm = llm,
    memory = memory,
    verbose = True
)
```

```python
conversations.predict(input="Hi,I'm Jasmine")
```

```python
conversations.predict(input="What is 1+1")
```

```python
conversations.predict(input="What is my name?")
```

```python
from langchain_classic.memory import ConversationTokenBufferMemory
# from langchain_community.llms import OpenAI
# 另一种会话记忆方式: token数量记忆，只保留最近max_token_limit个token以内的对话
# 注意: 旧教程此处会 from langchain_classic.llms import OpenAI，该写法已不可用——
# 旧版OpenAI类实际存放在langchain_community包中(本环境未安装，会报ModuleNotFoundError)，
# 且它调用的是旧版文本补全接口，DeepSeek并不支持；
# ConversationTokenBufferMemory中的llm仅用于统计token数量，复用上方已创建的ChatOpenAI实例即可

llm = ChatOpenAI(
    temperature=0,
    base_url="https://api.deepseek.com",
    model="deepseek-v4-flash",
    api_key=api_key,
    # 由于token计数的不同需要配置一个计数方式
    tiktoken_model_name="gpt-4o-mini"
)
```

```python
memory = ConversationTokenBufferMemory(llm=llm, max_token_limit=25)
memory.save_context({"input":"AI is what?!"},{"output":"Amazing!"})
memory.save_context({"input":"Backpropagation is what?"},{"output":"beautiful"})
memory.save_context({"input":"Chatbots are what?"},{"output":"Charming"})
```

```python
memory.load_memory_variables({})
```

```python
from langchain_classic.memory import ConversationSummaryBufferMemory # 让LLM为所有历史消息生成摘要，在记忆中存储历史消息的摘要
# create a long string
schedule = "There is a meeting at 8am with your product team. \
You will need your powerpoint presentation prepared. \
9am-12pm have time to work on your LangChain \
project which will go quickly because Langchain is such a po\
At Noon, lunch at the italian restaurant with a customer who \
from over an hour away to meet you to understand the last i\
Be sure to bring your laptop to show the latest LLM demo."

memory = ConversationSummaryBufferMemory(llm=llm,max_token_limit=400)
memory.save_context({"input":"Hi"},{"output":"What's up"}) # 加入输入、输出
memory.save_context({"input":"not much, just hanging"},
                    {"output": "Cool"})
memory.save_context({"input":"what is on the schedule today?"},{"output":f"{schedule}"})
```

```python
memory.load_memory_variables({})
```

```python
conversations = ConversationChain(
    llm=llm,
    memory=memory,
    verbose=True
)
```

```python
conversations.predict(input="What would be a good demo to show?")
```
